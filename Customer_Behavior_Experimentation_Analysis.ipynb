# Experiment Design in Online Marketing

## Question
Select an online business and develop an experiment, ensuring you adhere to each of the required steps.

## Answer

For this experiment, I selected **Uber**, an online ride-sharing platform. The business objective is to improve **driver retention**, since retaining active drivers is important for maintaining ride availability, reducing supply shortages, and supporting platform growth.

---

## 1. Define the Goal and Research Question

**Goal**

The goal of this experiment is to determine whether a new incentive program can improve driver retention.

**Research Question**

Will introducing the **“Power Hour”** feature, where drivers receive a **$25 bonus for completing 5 trips within a 2-hour period**, increase driver retention compared to the current system without this incentive?

---

## 2. Formulate the Hypothesis

**Null Hypothesis (H₀):**  
The Power Hour bonus does **not** increase driver retention compared to the existing system.

**Alternative Hypothesis (H₁):**  
The Power Hour bonus **increases** driver retention compared to the existing system.

This is a **one-tailed test** because the business objective is specifically to determine whether the incentive improves retention.

---

## 3. Design the Experiment (Pre-test Setup)

### 3.1 Determine Key Parameters

**Primary Metric**

Driver retention rate.

**Secondary Metrics**

- Number of trips completed per driver  
- Average driver earnings per hour  
- Driver app engagement during peak periods  

**Significance Level**

\[
\alpha = 0.05
\]

**Statistical Power**

\[
1 - \beta = 0.80
\]

**Minimum Detectable Effect (MDE)**

A meaningful increase in retention, for example **3–5% improvement** over the baseline retention rate.

---

### 3.2 Calculate Sample Size and Duration

The sample size should be large enough to detect the selected MDE with **80% statistical power** at the **5% significance level**.

**Planned Test Duration**

The experiment will run for **two weeks** to capture driver behavior across weekdays and weekends.

---

## 4. Implement the Test

Drivers will be randomly assigned into two groups:

**Control Group (A)**  
Drivers continue using the existing system without the Power Hour incentive.

**Treatment Group (B)**  
Drivers receive the Power Hour incentive where they earn a $25 bonus for completing 5 trips within 2 hours.

**Randomization**

Drivers will be randomly assigned to ensure both groups are statistically comparable and representative of the driver population.

**Isolation**

The feature will only be visible to drivers in the treatment group to avoid contamination between groups.

**Technical Setup**

Uber’s system will track trip completion, bonus qualification, earnings, and retention metrics for both groups.

---

## 5. Run the Experiment

The experiment will run for the full predetermined duration.

During the experiment, the team will monitor:

- Data quality
- Correct group assignment
- Tracking accuracy
- External events that may influence driver behavior

The experiment should **not be stopped early** to avoid misleading results caused by premature significance testing.

---

## 6. Analyze and Interpret Results

After the experiment concludes, retention rates between the two groups will be compared using a **statistical hypothesis test**, such as a **z-test for proportions**.

- If **p-value < 0.05**, the result is statistically significant and the null hypothesis is rejected.
- If **p-value ≥ 0.05**, the result is not statistically significant and the null hypothesis cannot be rejected.

A **confidence interval** will also be calculated to estimate the range of the true treatment effect.

Secondary metrics should also be evaluated to ensure there are no unintended negative outcomes such as driver fatigue or excessive operational costs.

---

## 7. Decision and Implementation

If the treatment group shows a **statistically significant improvement in driver retention**, Uber can roll out the Power Hour incentive to the full driver population.

If the results are **inconclusive**, the company may redesign the experiment by adjusting the bonus amount, trip requirement, or time window.

Finally, the experiment should be fully documented, including the hypothesis, experimental design, results, and final decision, to support future data-driven product development.

---

## Conclusion

This A/B test enables Uber to evaluate the effectiveness of the **Power Hour incentive** in improving driver retention. By using randomized experimentation and statistical analysis, Uber can make informed decisions about whether the feature should be implemented across the platform.

# I. Fundamental Statistical Concepts

## 1. What is the Central Limit Theorem and its significance?

The Central Limit Theorem (CLT) states that when the sample size is sufficiently large, the distribution of the sample mean approaches a normal distribution regardless of the original population distribution.

Its significance is that it allows analysts to apply statistical methods such as hypothesis testing and confidence intervals even when the population distribution is unknown.

---

## 2. Define sampling and list the sampling methods you are familiar with.

Sampling is the process of selecting a subset of observations from a larger population in order to make inferences about the population.

Common sampling methods include:

- Simple Random Sampling  
- Stratified Sampling  
- Cluster Sampling  
- Systematic Sampling  
- Convenience Sampling

Random sampling methods help reduce selection bias.

---

## 3. Explain the relationship between covariance and correlation.

Covariance measures how two variables change together. A positive covariance indicates that the variables move in the same direction, while a negative covariance indicates they move in opposite directions.

Correlation is a standardized version of covariance that ranges between -1 and +1, making it easier to interpret the strength and direction of the relationship between variables.

---

## 4. Define and explain the concept of a p-value in simple terms.

A p-value represents the probability of observing results at least as extreme as those obtained in the experiment assuming the null hypothesis is true.

If the p-value is smaller than the significance level (commonly 0.05), we reject the null hypothesis and conclude that the observed effect is statistically significant.

---

## 5. Differentiate between a null and an alternative hypothesis.

The **null hypothesis (H₀)** states that there is no effect or difference between groups.

The **alternative hypothesis (H₁)** states that there is a significant effect or difference.

Statistical tests evaluate whether there is sufficient evidence to reject the null hypothesis.

---

## 6. Explain the t-test.

A t-test is a statistical test used to determine whether the means of two groups are significantly different.

It is typically used when the population standard deviation is unknown and the sample size is relatively small.

---

## 7. What is the difference between a t-test and a Z-test?

A Z-test is used when the population variance is known and the sample size is large.

A t-test is used when the population variance is unknown and must be estimated from the sample, particularly when the sample size is small.

---

## 8. Highlight one key difference between one-tailed and two-tailed tests.

A one-tailed test evaluates whether a parameter is greater than or less than a specific value in a single direction.

A two-tailed test evaluates whether a parameter is different from a specific value in either direction.

---

## 9. What is the distinction between a Type I and a Type II error?

A Type I error occurs when a true null hypothesis is rejected (false positive).

A Type II error occurs when a false null hypothesis is not rejected (false negative).

The probability of a Type I error is denoted by α, while the probability of a Type II error is denoted by β.

---

# II. A/B Testing: Core Concepts and Metrics

## 1. Explain what A/B testing is and what its primary benefit is.

A/B testing is an experimental method used to compare two versions of a product, webpage, or feature to determine which performs better based on a predefined metric.

Its primary benefit is that it enables **data-driven decision making**, allowing companies to evaluate changes using controlled experiments.

---

## 2. Under what circumstances should an A/B test be run?

An A/B test should be run when a company wants to evaluate the impact of a specific change on a measurable outcome such as conversion rate, engagement, or retention.

---

## 3. What are the ideal conditions for A/B testing?

Ideal conditions include:

- Random assignment of users
- Sufficient sample size
- Independent observations
- Clearly defined success metrics
- Stable experimental conditions

---

## 4. Which parameters must be specified when setting up an A/B test?

Important parameters include:

- Primary metric
- Significance level (α)
- Statistical power (1 − β)
- Minimum Detectable Effect (MDE)
- Sample size
- Test duration

---

## 5. Explain the p-value specifically in the context of A/B testing.

In A/B testing, the p-value indicates the probability that the observed difference between the control and treatment groups occurred due to random chance.

If the p-value is less than the significance level (typically 0.05), the difference is considered statistically significant.

---

## 6. Explain the Confidence Interval in the context of A/B testing.

A confidence interval provides a range of values within which the true treatment effect is likely to lie with a specified level of confidence (commonly 95%).

It helps quantify both the magnitude of the effect and the uncertainty around the estimate.

---

## 7. How are alpha (α) and beta (β) defined in terms of A/B testing?

Alpha (α) represents the probability of a Type I error, meaning falsely detecting an effect when none exists.

Beta (β) represents the probability of a Type II error, meaning failing to detect a real effect.

Statistical power is defined as \(1 - \beta\).

---

## 8. What criteria determine what should be tested?

Features or changes that are likely to impact key business metrics should be prioritized for testing. These may include changes to user interface design, pricing strategies, recommendation systems, or incentive structures.

---

# Exercise 3: Causality with DAGs

For each directed acyclic graph (DAG), we:

1. Name the DAG
2. Generate a population dataset
3. Provide a marketing narrative aligned with the DAG
4. Compute the Average Treatment Effect (ATE) on the population
5. Estimate the ATE using a sample of 10,000 observations

The treatment variable is **X**, and the outcome variable is **Y**.

```python
import numpy as np
import pandas as pd

np.random.seed(42)
def compute_population_ate(df):
    return df[df["X"]==1]["Y"].mean() - df[df["X"]==0]["Y"].mean()

def compute_sample_ate(df, sample_size=10000):
    sample = df.sample(sample_size, random_state=42)
    return sample[sample["X"]==1]["Y"].mean() - sample[sample["X"]==0]["Y"].mean()
```

## DAG 1: Mediator

Structure: X → M → Y

Marketing Narrative:

An online company launches a digital advertising campaign (X).  
The campaign increases brand awareness (M).  
Greater awareness leads to higher product purchases (Y).

Here, **awareness acts as a mediator** between advertising and sales.

```python
def mediator_population(n=200000):

    X = np.random.binomial(1,0.5,n)

    M = 0.8*X + np.random.normal(0,1,n)

    Y = 0.5*X + 1.2*M + np.random.normal(0,1,n)

    return pd.DataFrame({"X":X,"M":M,"Y":Y})

df_mediator = mediator_population()

print("Mediator Population ATE:", compute_population_ate(df_mediator))
print("Mediator Sample ATE:", compute_sample_ate(df_mediator))
```

## DAG 2: Confounder

Structure: C → X and C → Y

Marketing Narrative:

Customer loyalty (C) influences both whether a customer receives a promotional email (X) and how much they spend (Y).  
If loyalty is not controlled for, the estimated treatment effect may be biased.

```python
def confounder_population(n=200000):

    C = np.random.normal(0,1,n)

    X = (C + np.random.normal(0,1,n) > 0).astype(int)

    Y = 0.6*X + 1.0*C + np.random.normal(0,1,n)

    return pd.DataFrame({"C":C,"X":X,"Y":Y})

df_confounder = confounder_population()

print("Confounder Population ATE:", compute_population_ate(df_confounder))
print("Confounder Sample ATE:", compute_sample_ate(df_confounder))
```

## DAG 3: Collider

Structure: X → C ← Y

Marketing Narrative:

Website traffic source (X) and customer purchase intent (Y) both influence whether a user lands on a special campaign page (C).  
Conditioning on C can create a false association between X and Y.

```python
def collider_population(n=200000):

    X = np.random.normal(0,1,n)

    Y = 0.7*X + np.random.normal(0,1,n)

    C = X + Y + np.random.normal(0,1,n)

    X_binary = (X>0).astype(int)

    return pd.DataFrame({"X":X_binary,"Y":Y,"C":C})

df_collider = collider_population()

print("Collider Population ATE:", compute_population_ate(df_collider))
print("Collider Sample ATE:", compute_sample_ate(df_collider))
```

## DAG 4: Moderator

Structure: M modifies the strength of X → Y

Marketing Narrative:

An email promotion (X) affects purchases (Y), but the effect depends on customer engagement level (M).  
Highly engaged customers respond more strongly.

```python
def moderator_population(n=200000):

    X = np.random.binomial(1,0.5,n)

    M = np.random.binomial(1,0.5,n)

    Y = 0.5*X + 1.0*M + 1.2*X*M + np.random.normal(0,1,n)

    return pd.DataFrame({"X":X,"M":M,"Y":Y})

df_moderator = moderator_population()

print("Moderator Population ATE:", compute_population_ate(df_moderator))
print("Moderator Sample ATE:", compute_sample_ate(df_moderator))
```

## DAG 5: Instrumental Variable

Structure: Z → X → Y

Marketing Narrative:

A random notification (Z) encourages users to open an app (X).  
Opening the app increases purchases (Y).  
The notification does not directly affect purchases.

```python
def iv_population(n=200000):

    Z = np.random.binomial(1,0.5,n)

    X = (0.8*Z + np.random.normal(0,1,n) > 0).astype(int)

    Y = 0.7*X + np.random.normal(0,1,n)

    return pd.DataFrame({"Z":Z,"X":X,"Y":Y})

df_iv = iv_population()

print("IV Population ATE:", compute_population_ate(df_iv))
print("IV Sample ATE:", compute_sample_ate(df_iv))
```

## DAG 6: M-bias

Structure: U1 → X, U1 → C, U2 → C, U2 → Y

Marketing Narrative:

Two hidden variables influence both treatment exposure and outcomes.  
Controlling for the wrong variable (C) can introduce bias.

```python
def mbias_population(n=200000):

    U1 = np.random.normal(0,1,n)

    U2 = np.random.normal(0,1,n)

    X = (U1 + np.random.normal(0,1,n) > 0).astype(int)

    C = U1 + U2 + np.random.normal(0,1,n)

    Y = 0.6*X + U2 + np.random.normal(0,1,n)

    return pd.DataFrame({"U1":U1,"U2":U2,"X":X,"C":C,"Y":Y})

df_mbias = mbias_population()

print("M-Bias Population ATE:", compute_population_ate(df_mbias))
print("M-Bias Sample ATE:", compute_sample_ate(df_mbias))
```

## DAG 7: Front-Door Path

Structure: X → M → Y

Marketing Narrative:

Online advertisements (X) influence click behavior (M), which then drives purchases (Y).  
Even if unobserved factors affect both X and Y, observing M allows causal identification.

```python
def frontdoor_population(n=200000):

    X = np.random.binomial(1,0.5,n)

    M = 1.0*X + np.random.normal(0,1,n)

    Y = 1.2*M + np.random.normal(0,1,n)

    return pd.DataFrame({"X":X,"M":M,"Y":Y})

df_frontdoor = frontdoor_population()

print("Front-Door Population ATE:", compute_population_ate(df_frontdoor))
print("Front-Door Sample ATE:", compute_sample_ate(df_frontdoor))
```

## DAG 8: Selection Bias

Structure: X → S ← Y

Marketing Narrative:

A company evaluates a marketing campaign (X) using only customers who completed purchases (S).  
Since both exposure and purchase behavior influence selection, the sample may produce biased conclusions.

```python
def selection_population(n=200000):

    X = np.random.normal(0,1,n)

    Y = 0.6*X + np.random.normal(0,1,n)

    S = X + Y + np.random.normal(0,1,n)

    X_binary = (X>0).astype(int)

    return pd.DataFrame({"X":X_binary,"Y":Y,"S":S})

df_selection = selection_population()

print("Selection Population ATE:", compute_population_ate(df_selection))
print("Selection Sample ATE:", compute_sample_ate(df_selection))
```

## Interpretation

The sample ATE uses only 10,000 observations to verify that the estimated effect is close to the population value.

Small differences between the two estimates occur due to sampling variability, but the results should generally be similar when the sample size is sufficiently large.

# Exercise 4: Booking Cancellation

This analysis investigates the causal factors behind hotel booking cancellations using a simulated observational dataset.

The objective is to evaluate how different observable variables affect the probability that a booking is cancelled.

The analysis follows four steps:

1. Generate a population dataset of 3,000,000 observations.
2. Compute the true population Average Treatment Effect (ATE).
3. Draw a sample of 10,000 observations and estimate the ATE using regression.
4. Repeat the sampling and estimation process 1,000 times and plot the distribution of estimated effects.

The outcome variable is **booking_cancelled**.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

np.random.seed(42)
N = 3000000

lead_time = np.random.exponential(30, N)
deposit = np.random.binomial(1,0.2,N)
previous_cancellations = np.random.poisson(0.5,N)
special_requests = np.random.poisson(1,N)
adr = np.random.normal(100,20,N)

# nonlinear cancellation probability
logit = (
    0.02*lead_time
    + 0.8*deposit
    + 0.6*np.sqrt(previous_cancellations)
    - 0.5*np.log1p(special_requests)
    + 0.01*(adr-100)**2/100
)

prob_cancel = 1/(1+np.exp(-logit))

booking_cancelled = np.random.binomial(1,prob_cancel)

population = pd.DataFrame({
    "lead_time":lead_time,
    "deposit":deposit,
    "previous_cancellations":previous_cancellations,
    "special_requests":special_requests,
    "adr":adr,
    "booking_cancelled":booking_cancelled
})

population.head()
```

## Step 2: True Population ATE

The Average Treatment Effect (ATE) measures the effect of each observable variable on the probability of booking cancellation in the full population.

```python
variables = ["lead_time","deposit","previous_cancellations","special_requests","adr"]

population_ate = {}

for var in variables:

    high = population[var] > population[var].median()

    ate = population[high]["booking_cancelled"].mean() - population[~high]["booking_cancelled"].mean()

    population_ate[var] = ate

population_ate
```

## Step 3: Sample Estimation (n = 10,000)

A random sample of 10,000 observations is drawn from the population.

Regression is used to estimate the effect of each variable on booking cancellations.

```python
sample = population.sample(10000,random_state=42)

X = sample[variables]
X = sm.add_constant(X)

y = sample["booking_cancelled"]

model = sm.OLS(y,X).fit()

print(model.summary())
```

## Step 4: Replication (1,000 simulations)

We repeat the sampling and regression process 1,000 times.

This allows us to observe how the estimated effects vary due to sampling variation.

```python
import numpy as np

# use a smaller working population for faster sampling
working_population = population.sample(300000, random_state=42)

estimates = {var: [] for var in variables}

for i in range(300):   # reduced from 1000 to run faster

    sample = working_population.sample(10000)

    X = sample[variables].values
    X = np.column_stack((np.ones(len(X)), X))   # add intercept

    y = sample["booking_cancelled"].values

    beta = np.linalg.inv(X.T @ X) @ (X.T @ y)

    for j, var in enumerate(variables):
        estimates[var].append(beta[j+1])
```

```python
fig, axes = plt.subplots(3,2, figsize=(12,10))
axes = axes.flatten()

for i, var in enumerate(variables):
    axes[i].hist(estimates[var], bins=30)
    axes[i].axvline(population_ate[var], color="red", label="Population ATE")
    axes[i].set_title(var)
    axes[i].legend()

axes[-1].axis('off')  # remove empty plot

plt.tight_layout()
plt.show()
```

## Interpretation of Results

The histograms show the distribution of estimated effects obtained from repeated samples of size 10,000.

The red vertical line represents the true population Average Treatment Effect (ATE) computed using the full simulated population of 3 million observations.

Most of the estimated effects cluster around the population ATE, indicating that the regression estimator is approximately unbiased. However, there is still variability in the estimates due to sampling variation.

This demonstrates an important concept in causal inference: even when the true causal effect is fixed, estimates from finite samples may vary. As the sample size increases, the estimates tend to converge closer to the true population effect.
